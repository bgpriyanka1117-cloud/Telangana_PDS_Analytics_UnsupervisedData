```python
df.drop_duplicates(inplace=True)

print("Duplicate rows removed")
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df['fpsType'] = df['fpsType'].fillna('Inactive Shop')
```

```python
df['fpsType'].value_counts(dropna=False)
```

```python
df.drop(columns=['day'], inplace=True)
```

```python
'day' in df.columns
```

```python
df['fpsStatus'] = df['fpsStatus'].fillna('Inactive')
```

```python
df['fpsStatus'].value_counts(dropna=False)
```

```python
df['distCode'] = df['distCode'].fillna(df['distCode_x'])
```

```python
df['distCode'].isnull().sum()
```

```python
df['distName'] = df['distName'].fillna(df['distName_x'])
```

```python
df['distName'].isnull().sum()
```

```python
df['officeCode'] = df['officeCode'].fillna(df['officeCode_x'])
```

```python
df['officeCode'].isnull().sum()
```

```python
df['officeName'] = df['officeName'].fillna(df['officeName_x'])
```

```python
df['officeName'].isnull().sum()
```

```python
drop_cols = [
    'distCode_x',
    'distName_x',
    'officeCode_x',
    'officeName_x',
    'distCode_y',
    'distName_y',
    'officeCode_y',
    'officeName_y'
]

df.drop(columns=drop_cols, inplace=True)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
print(df[['year_x', 'year_y']].head())

print(df[['month_x', 'month_y']].head())
```

```python
(df['year_x'] == df['year_y']).value_counts(dropna=False)
```

```python
(df['month_x'] == df['month_y']).value_counts(dropna=False)
```

```python
df.rename(columns={
    'year_x': 'year',
    'month_x': 'month'
}, inplace=True)
```

```python
df['dateTime'] = pd.to_datetime(df['dateTime'], errors='coerce')
```

```python
df.drop(columns=['address'], inplace=True)
```

```python
df[['latitude','longitude']].describe()
```

```python
df['latitude'] = df['latitude'].fillna(
    df.groupby('shopNo')['latitude'].transform('median')
)

df['longitude'] = df['longitude'].fillna(
    df.groupby('shopNo')['longitude'].transform('median')
)
```

```python
df['latitude'] = df['latitude'].fillna(df['latitude'].median())
df['longitude'] = df['longitude'].fillna(df['longitude'].median())
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.drop(columns=['month_y', 'year_y'], inplace=True)
```

```python
df.rename(columns={
    'month_x': 'month',
    'year_x': 'year'
}, inplace=True)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df['dateTime'] = df['dateTime'].fillna(method='ffill')
```

```python
df.isnull().sum().sort_values(ascending=False)
```
