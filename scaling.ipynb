```python
import pandas as pd
import numpy as np
```

```python
#Load dataset
df=pd.read_csv("../data/Master_data/Encoding_data.csv")
print(df.head)
```

```python
df.shape
```

```python
df.columns
```

```python
#These are IDs / identifiers.
drop_cols = [
    'shopNo',
    'distCode',
    'officeCode'
]

# Drop columns
df = df.drop(columns=drop_cols, errors='ignore')

# Check remaining columns
print(df.shape)
df.head()
```

```python
df.shape
```

```python
for col in drop_cols:
    print(col, "→", col in df.columns)
```

```python
#step 2:-These are useful for seasonality patterns.
time_cols = [
    'month',
    'year',
    'week',
    'quarter',
    'day_of_year',
    'days_in_month'
]
```

```python
#scaling Times..
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

df[time_cols] = scaler.fit_transform(df[time_cols])

df[time_cols].head()
```

```python
#Useful for regional clustering.
geo_cols = [
    'longitude',
    'latitude'
]
```

```python
from sklearn.preprocessing import StandardScaler


scaler = StandardScaler()

df[geo_cols] = scaler.fit_transform(df[geo_cols])

df[geo_cols].head()
```

```python
df[['longitude', 'latitude']].describe().T
```

```python
# CONTINUOUS NUMERIC FEATURES (Scale)
continuous_cols = [
    'noOfRcs',
    'noOfTrans',
    'riceAfsc',
    'riceFsc',
    'riceAap',
    'wheat',
    'sugar',
    'totalAmount',
    'otherShopTransCnt',
    'riceAap_available',
    'wheat_available',
    'totalRcs',
    'totalUnits',
    'units',
    'utilization_ratio',
    'units_per_card',
    'amount_per_transaction',
    'amount_per_card',
    'rice_afsc_per_trans',
    'rice_fsc_per_trans',
    'rice_aap_per_trans',
    'wheat_per_trans',
    'rice_to_wheat_ratio',
    'sugar_intensity',
    'wheat_intensity',
    'rice_share',
    'commodity_activity',
    'other_shop_ratio',
    'portability_hub_score',
    'portability_intensity',
    'trans_per_card_FRAUD_DETECTION',
    'units_per_trans_FRAUD_DETECTION',
    'distribution_change',
    'transaction_volatility',
    'amount_volatility',
    'commodity_volatility',
    'amount_growth',
    'commodity_growth',
    'activity_density',
    'shop_density',
    'district_load',
    'seasonal_demand',
    'availability_score',
    'rice_wheat_ratio',
    'distName_freq'
]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

df[continuous_cols] = scaler.fit_transform(
    df[continuous_cols]
)

df[continuous_cols].head()
```

```python
df.describe().T[['mean', 'std', 'min', 'max']]
```

```python
df.to_csv("../data/Master_data/scaled_data.csv",index=False)
```
