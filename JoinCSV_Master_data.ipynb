```python
import pandas as pd
```

```python
card_master = pd.read_csv("../data/clean_data_2023_2025/clean_card_data.csv")
print(card_master.shape)
card_master.columns
```

```python
shop_master = pd.read_csv("../data/clean_data_2023_2025/clean_Shop_data.csv")
print(shop_master.shape)
shop_master.columns
```

```python
trans_master = pd.read_csv("../data/clean_data_2023_2025/clean_transaction_data.csv")
print(trans_master.shape)
trans_master.columns
```

```python
trans_master.year.value_counts()
```

```python
print(card_master.head())

print(shop_master.head())

print(trans_master.head())
```

```python
print("Card Rows :", len(card_master))

print("Shop Rows :", len(shop_master))

print("Transaction Rows :", len(trans_master))
```

```python
print(
    card_master.groupby(
        ['shopNo','distCode']
    ).size().reset_index(name='count')
)

print(
    trans_master.groupby(
        ['shopNo','distCode']
    ).size().reset_index(name='count')
)

print(
    shop_master.groupby(
        ['shopNo','distCode']
    ).size().reset_index(name='count')
)
```

```python
trans_master.columns
```

```python
print(
    trans_master.groupby(
        ['shopNo','year','month']
    ).agg({'sugar':'sum'})
)
```

```python
card_master = card_master.drop_duplicates(
    subset=['shopNo','distCode','month','year']
)

trans_master = trans_master.drop_duplicates(
    subset=['shopNo','distCode','month','year']
)
```

```python
shop_master = shop_master.drop_duplicates(
    subset=['shopNo','distCode']
)
```

```python
print(len(card_master))

print(len(trans_master))

print(len(shop_master))
```

```python
merged_df = pd.merge(
    trans_master,
    card_master,
      on=['shopNo','month','year'],
    how='inner'
)
```

```python
final_df = pd.merge(
    merged_df,
    shop_master,
    on=['shopNo'],
    how='left'
)
```
