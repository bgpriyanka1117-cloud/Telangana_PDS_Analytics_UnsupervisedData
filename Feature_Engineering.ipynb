```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('../data/Master_data/outlier_cliping_dataset.csv')
```

```python
df.shape
```

```python
df.columns
```

```python
# check Numeric or Categorical coloumns
print(df.select_dtypes(include='object').columns)
numeric_cols = df.select_dtypes(include=np.number).columns

print(numeric_cols)
```

```python
# DATE & TIME FEATURES
df['dateTime'] = pd.to_datetime(df['dateTime'])

df['day_name'] = df['dateTime'].dt.day_name()
df['week'] = df['dateTime'].dt.isocalendar().week
df['quarter'] = df['dateTime'].dt.quarter
df['is_weekend'] = (df['dateTime'].dt.weekday >= 5).astype(int)
df['day_of_year'] = df['dateTime'].dt.dayofyear
df['month_start'] = df['dateTime'].dt.is_month_start.astype(int)
df['month_end'] = df['dateTime'].dt.is_month_end.astype(int)
df['days_in_month'] = df['dateTime'].dt.days_in_month
```

```python
#BASIC UTILIZATION FEATURES

# Transactions per ration card
df['utilization_ratio'] = (
    df['noOfTrans'] /
    (df['totalRcs'] + 1)
)
df
```

```python
# Units distributed per ration card
df['units_per_card'] = (
    df['totalUnits'] /
    (df['totalRcs'] + 1)
 )
df
```

```python
# Average transaction amount
df['amount_per_transaction'] = (
    df['totalAmount'] /
    (df['noOfTrans'] + 1)
)
df
```

```python
# Average transaction value per card
df['amount_per_card'] = (
    df['totalAmount'] /
    (df['totalRcs'] + 1)
)
df
```

```python
# COMMODITY FEATURES

# Commodity per transaction
df['rice_afsc_per_trans'] = (
    df['riceAfsc'] /
    (df['noOfTrans'] + 1)
)
df
```

```python
df['rice_fsc_per_trans'] = (
    df['riceFsc'] /
    (df['noOfTrans'] + 1)
)
df
```

```python
df['rice_aap_per_trans'] = (
    df['riceAap'] /
    (df['noOfTrans'] + 1)
)
df
```

```python
df['wheat_per_trans'] = (
    df['wheat'] /
    (df['noOfTrans'] + 1)
)
df
```

```python
# COMMODITY INTENSITY FEATURES
# Rice-to-Wheat Ratio
df['rice_to_wheat_ratio'] = (
    (df['riceAfsc'] + df['riceFsc'] + df['riceAap']) /
    (df['wheat'] + 1)
)
df
```

```python
# Sugar Intensity
df['sugar_intensity'] = (
    df['sugar'] /
    (df['totalUnits'] + 1)
)
df
```

```python
# Wheat Intensity
df['wheat_intensity'] = (
    df['wheat'] /
    (df['totalUnits'] + 1)
)
df
```

```python
# Rice Share
df['rice_share'] = (
    (df['riceAfsc'] + df['riceFsc'] + df['riceAap']) /
    (df['totalUnits'] + 1)
)
df
```

```python
# Total Commodity Activity
df['commodity_activity'] = (
    (df['riceAfsc'] > 0).astype(int) +
    (df['riceFsc'] > 0).astype(int) +
    (df['riceAap'] > 0).astype(int) +
    (df['wheat'] > 0).astype(int) +
    (df['sugar'] > 0).astype(int)
)
df
```

```python
# PORTABILITY FEATURES
# Other shop dependency
df['other_shop_ratio'] = (
    df['otherShopTransCnt'] /
    (df['noOfTrans'] + 1)
)
df
```

```python
# Portability hub score
df['portability_hub_score'] = (
    df['otherShopTransCnt'] *
    df['utilization_ratio']
)
df
```

```python
# Portability intensity
df['portability_intensity'] = (
    df['otherShopTransCnt'] /
    (df['totalRcs'] + 1)
)
df
```

```python
# High non-local traffic shops
df['high_non_local_flag'] = (
    df['other_shop_ratio'] >
    df['other_shop_ratio'].quantile(0.90)
).astype(int)
df
```

```python
#FRAUD DETECTION FEATURES
# Transactions per card
df['trans_per_card_FRAUD_DETECTION'] = (
    df['noOfTrans'] /
    (df['totalRcs'] + 1)
)
df
```

```python
# Units per transaction
df['units_per_trans_FRAUD_DETECTION'] = (
    df['totalUnits'] /
    (df['noOfTrans'] + 1)
)
df
```

```python
print(df.columns)
```

```python
# Fraud risk threshold
fraud_threshold = (
    df['trans_per_card_FRAUD_DETECTION']
    .quantile(0.99)
)
df
```

```python
# Fraud risk flag
df['fraud_risk_flag'] = (
    df['trans_per_card_FRAUD_DETECTION'] >
    fraud_threshold
).astype(int)

df
```

```python
# POLICY IMPACT FEATURES
# ONORC implementation indicator
df['post_onorc'] = (
    df['year'] >= 2020
).astype(int)
df
```

```python
# Urban/Rural Proxy
office_counts = (
    df.groupby('distCode')['officeCode']
    .transform('count')
)
df
```

```python
df['urban_proxy'] = (
    office_counts >
    office_counts.median()
).astype(int)
df
```

```python
# Policy impact feature
df['distribution_change'] = (
    df['post_onorc'] *
    df['other_shop_ratio']
)
df
```

```python
# VOLATILITY FEATURES
# Transaction volatility
df['transaction_volatility'] = (
    df.groupby('shopNo')['noOfTrans']
    .transform('std')
)
df
```

```python
# Amount volatility
df['amount_volatility'] = (
    df.groupby('shopNo')['totalAmount']
    .transform('std')
)
df
```

```python
# Commodity volatility
df['commodity_volatility'] = (
    df.groupby('shopNo')['totalUnits']
    .transform('std')
)
df
```

```python
# GROWTH FEATURES
df = df.sort_values(
    ['shopNo', 'year', 'month']
)
df
```

```python
df.isnull().sum()
```

```python
# Amount growth
df['amount_growth'] = df['totalAmount'] / df['totalRcs']
df
```

```python
df['amount_growth'].unique()
```

```python
# Commodity growth
df['commodity_growth'] = (
    df.groupby('shopNo')['totalUnits']
    .pct_change()
)
df
```

```python
df['commodity_growth'] = (
    df['commodity_growth']
    .fillna(0)
)
```

```python
df['commodity_growth'].isnull().sum()
```

```python
# GEO FEATURES

# Average district activity
df['activity_density'] = (
    df.groupby('distCode')['noOfTrans']
    .transform('mean')
)
df
```

```python
# Shop density
df['shop_density'] = (
    df.groupby('distCode')['shopNo']
    .transform('nunique')
)
df
```

```python
# District transaction load
df['district_load'] = (
    df.groupby('distCode')['totalAmount']
    .transform('mean')
)
df
```

```python
# SEASONAL FEATURES
# Seasonal demand
df['seasonal_demand'] = (
    df.groupby('month')['noOfTrans']
    .transform('mean')
)
df
```

```python
# Peak demand month
peak_threshold = (
    df['seasonal_demand']
    .quantile(0.90)
)
df
```

```python
df['peak_month_flag'] = (
    df['seasonal_demand'] >
    peak_threshold
).astype(int)
df
```

```python
df['peak_month_flag'] = (
    df['seasonal_demand'] >
    peak_threshold
).astype(int)
df
```

```python
# Festival quarter flag
df['festival_quarter'] = (
    df['quarter'].isin([3, 4])
).astype(int)

df
```

```python
# SMART SHOP FEATURES
# High performance shops
high_perf_threshold = (
    df['utilization_ratio']
    .quantile(0.90)
)
df['high_performance_shop'] = (
    df['utilization_ratio'] >
    high_perf_threshold
).astype(int)
df
```

```python
# Low activity shops
low_activity_threshold = (
    df['utilization_ratio']
    .quantile(0.10)
)

df['low_activity_shop']=(
    df['utilization_ratio']<
    low_activity_threshold
).astype(int)
df
```

```python
# High revenue shops
high_revenue_threshold=(
    df['totalAmount'].quantile(0.90)
)

df['high_revenue_shop'] = (
    df['totalAmount'] >
    high_revenue_threshold
).astype(int)
df
```

```python
# AVAILABILITY FEATURES
# Commodity availability score
df['availability_score'] = (
    df['riceAap_available'] +
    df['wheat_available']
)
df
```

```python
# STATUS FEATURES
# Active FPS indicator
df['active_fps'] = (
    df['fpsStatus']
    .astype(str)
    .str.lower()
    .isin(['active', 'running'])
).astype(int)

df
```

```python
 #LOCATION FEATURES
# Missing geo flag
df['missing_geo_flag'] = (
    df['longitude'].isnull() |
    df['latitude'].isnull()
).astype(int)
df
```
