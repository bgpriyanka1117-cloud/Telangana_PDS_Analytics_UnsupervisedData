```python
import pandas as pd 
#Geo Profiling Label
#Urban/Rural Label
#Fraud Risk Label
#Business Cluster Label
```

```python
main_df=pd.read_csv("../data/Final_meaningful_dataset_Clusting.csv")
print(main_df.head())
```

```python
main_df.shape
```

```python
# ==========================================
# CLUSTER LABELING TO MAIN DATAFRAME
# ==========================================

# GEO LABEL
geo_labels = {
    0: "Urban High Density Cluster",
    1: "Semi-Urban Cluster",
    2: "Rural Cluster",
    3: "Urban High Density Cluster"
}

# URBAN / RURAL LABEL
area_labels = {
    0: "Urban",
    1: "Semi-Urban",
    2: "Rural",
    3: "Urban"
}

# FRAUD RISK LABEL
fraud_labels = {
    0: "Low Fraud Risk Cluster",
    1: "Medium Fraud Risk Cluster",
    2: "High Fraud Risk Cluster",
    3: "Low Fraud Risk Cluster"
}

# FRAUD DESCRIPTION
fraud_desc = {
    0: "Stable behavior | Low anomaly risk",
    1: "Monitor regularly | Moderate anomaly signals",
    2: "Suspicious activity detected | High volatility | Requires audit",
    3: "Stable behavior | Low anomaly risk"
}

# BUSINESS CLUSTER LABEL
business_labels = {
    0: "Low Activity Rural FPS",
    1: "Stable Semi-Urban FPS",
    2: "High Value Commercial Hub",
    3: "Urban High Performance FPS"
}

# BUSINESS DESCRIPTION
business_desc = {
    0: "Underutilized shops | Low transactions | Weak demand network",
    1: "Medium activity | Balanced operations | Stable distribution system",
    2: "Very high transactions | High revenue system | Critical supply chain nodes",
    3: "Efficient urban system | Stable demand-supply balance | Low risk environment"
}

# ==========================================
# APPLY LABELS TO MAIN DF
# ==========================================

main_df["Geo_Profile"] = (
    main_df["Cluster"]
    .map(geo_labels)
)

main_df["Area_Type"] = (
    main_df["Cluster"]
     .map(area_labels)
)

main_df["Fraud_Risk"] = (
    main_df["Cluster"]
    .map(fraud_labels)
)

main_df["Fraud_Description"] = (
    main_df["Cluster"]
    .map(fraud_desc)
)

main_df["Business_Profile"] = (
    main_df["Cluster"]
    .map(business_labels)
)

main_df["Business_Description"] = (
    main_df["Cluster"]
    .map(business_desc)
)

# ==========================================
# CHECK RESULT
# ==========================================
print(main_df.head())
```

```python
main_df.to_csv( "../data/Final_Cluster_Lables_Profile.csv"
)

print("Cluster Labled saved successfully")
```

```python
main_df.to_pickle( "../app/Final_Cluster_Lables_Profile.pkl"
)
print("Cluster Labled saved successfully")
```
